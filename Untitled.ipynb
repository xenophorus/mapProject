region = "Приморский край, Россия"
primorye = ox.features_from_place(
    region, 
    tags = {"boundary": "administrative", "admin_level": "6"}
)
primorye.to_file("primorye.geojson", driver="GeoJSON")
city_name = "Владивосток, Приморский край, Россия"

vlad_districts.to_file("vladivostok.geojson", driver="GeoJSON")
